Grafico de Barras

```python
meses = ['Enero', 'Febrero', 'Marzo']
ventas = [1500, 1700, 1200]
plt.figure(figsize=(10,7))
plt.bar(meses, ventas, color='Green') ### Es el tipo de grafico
plt.title('Ventas por mes', fontsize= 12)
plt.xlabel('Meses', fontsize= 12)
plt.ylabel('Ventas', fontsize= 12)
### etiquetas de datos
for i in range(len(meses)):
  plt.text(meses[i], ventas[i] + 50, str(ventas[i]), ha='center', fontsize=12)
plt.show()
```

```python
matriz = np.random.randint(1,101,(100,100))
### crear una figura y un eje
fig, grafico = plt.subplots()
## mostrar mapa de color en la matriz
cgrafico = grafico.imshow(matriz, cmap='viridis', interpolation='nearest') ###interpolation genera los puntos
fig.colorbar(cgrafico)
###añadir titulo y etiquetas
grafico.set_title('Mapa de color')
grafico.set_xlabel('Indice de la columna')
grafico.set_ylabel('Indice de la fila')
plt.show()
```
