```python
import numpy as np
```

creando un grafico simple

```python
vector = np.random.randint(1,11,10)
```

```python
import matplotlib.pyplot as plt
plt.plot(vector, label="vector")
plt.title("Grafico de Lineas, vector")
plt.xlabel("secuencia")
plt.ylabel("valores")

plt.grid(True)
plt.legend()
plt.show()
```

un grafico de lineas pero mas elevado

```python
plt.plot(vector, marker="x", color="Green", linestyle= "--", label = "vector", markerfacecolor = "Green")
plt.xlabel("secuencia")
plt.ylabel("valores")

plt.grid(True)
plt.legend()
plt.show()
```

Haciendo un grafico de barras

```python
meses= ['abril', 'mayo', 'Junio']
ventas= [1600, 1300, 1000]
plt.figure(figsize=(5,5))
plt.bar(meses, ventas, color= 'Blue')
plt.title("Grafico de Barras", fontsize=12)
plt.xlabel("precio venta")
plt.ylabel("meses")

for i in range(len(meses)):
  plt.text(meses[i], ventas[i] + 30, str(ventas[i]), ha= 'center', fontsize = 12)
plt.show()
```

Mapa de color


```python
matriz = np.random.randint(1,101,(100,100))
fig, grafico = plt.subplots()
cgrafico= grafico.imshow(matriz, cmap = "viridis", interpolation= 'nearest')
fig.colorbar(cgrafico)
plt.title("mapa de color")
plt.xlabel("barra de color")
plt.ylabel("colores")
plt.show()
```
