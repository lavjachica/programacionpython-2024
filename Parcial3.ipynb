Este es para eliminar datos nulos, el anterior para mostrar las filas nulas
dropna elimina datos nulos del data set


```python
datos_sin_nulos= datos.dropna()
datos_sin_nulos.info() #### muestra la informacion
```

sensible a mayuscula y minuscula

```python
def filtro_rpg(genres):
  return 'RPG' in genres

# Se aplica el filtro utilizando apply
datosRPG = datos_sin_nulos[datos_sin_nulos['Genres'].apply(filtro_rpg)]
```

Transformacion de dataset: muestra 2 columnas de datos (solo juegos rpg)

```python
nombresRPG = datosRPG[['Title', 'Rating']]
nombresRPG.head(20) ## Head muestra los primeros 20 elementos
```

```python
nombresRPG.info()
```

```python
datasetRPG = nombresRPG.head(10)
datasetRPG.set_index('Title', inplace=True)
grafico = datasetRPG['Rating'].plot(x=None, kind='bar')
for i, rating in enumerate(datasetRPG['Rating']):
  grafico.annotate(str(rating), xy=(i, rating), ha='center', va='bottom')
plt.show
```
