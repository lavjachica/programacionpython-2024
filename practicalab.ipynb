1. Generar un vector aleatorio que incluya numeros negativos desde el -10 hasta el 10 positivo. Este vector debe tener 25 elementos. Deben ser numeros flotantes

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
vector = np.random.uniform(-10,10,25)
vector
```

2. Transformar el vector en una matriz cuadrada de 5x5. Obtener el promedio de la matriz

```python
matriz = vector.reshape(5,5)
matriz
```

```python
promedio = np.mean(matriz)
promedio
```

3. Generar un mapa de color a partir de la matriz generada anteriormente.

```python
fig, grafico = plt.subplots()
cgrafico = grafico.imshow(matriz, cmap='viridis', interpolation='nearest')
fig.colorbar(cgrafico)
grafico.set_title('Mapa de color')
grafico.set_xlabel('Indice de la columna')
grafico.set_ylabel('Indice de la fila')
plt.show()
```
